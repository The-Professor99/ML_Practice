```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib
```

```python
from sklearn.datasets import make_moons
```

```python
dataset = make_moons(n_samples=10000, noise=0.4)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(dataset[0], dataset[1], test_size=0.1)
```

```python
def plot_dataset(X, y):
    if X.shape[1] == 1:
        plt.plot(X, y, "bs")
    else:
        plt.plot(X[:, 0][y==0], X[:, 1][y==0], "bs")
        plt.plot(X[:, 0][y==1], X[:, 1][y==1], "g^")
    plt.grid(True, which='both')
    plt.xlabel(r"$x_1$", fontsize=20)
    plt.ylabel(r"$x_2$", fontsize=20, rotation=0)
```

```python
plot_dataset(X_train, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree_clf = DecisionTreeClassifier(max_depth=2)
```

```python
tree_clf.fit(X_train, y_train)
```

```python
from sklearn.tree import export_graphviz
```

```python
export_graphviz(tree_clf, out_file="images/exercise_tree.dot",
               class_names=["0", "1"], feature_names=["x1", "x2"])
```

```python
y_pred = tree_clf.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score
```

```python
accuracy_score(y_train, y_pred)
```

```python
recall_score(y_train, y_pred)
```

```python
precision_score(y_train, y_pred)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(tree_clf, X_train, y_train, cv=5)
```

```python
scores
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
param_grid = [
    {"max_depth": [2, 4, 6, 9, 15],
     "min_samples_split": [2, 40, 100, 500, 1000],
     "min_samples_leaf": [1, 5, 20, 30, 40],
     "max_leaf_nodes": [None, 10, 20, 30, 40, 50],
    }
]
```

```python
grid_search = GridSearchCV(tree_clf, param_grid, cv=5, return_train_score=True)
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
cv_res = grid_search.cv_results_
```

```python
for score, params in zip(cv_res["mean_test_score"], cv_res["params"]):
    print(score, params)
```

```python
final_clf = grid_search.best_estimator_
```

### Evaluating on a test set

```python
test_pred = final_clf.predict(X_test)
```

```python
accuracy_score(y_test, test_pred)
```

```python
precision_score(y_test, test_pred)
```

```python
recall_score(y_test, test_pred)
```

```python
n_trees = 1000
n_instances = 100

subsets = []
split = ShuffleSplit(n_splits=1000, test_size=len(X_train) - n_instances)
for train_index, test_index in split.split(X_train, y_train):
    shuffle_train_set = X_train[train_index]
    shuffle_test_set = y_train[train_index]
    subsets.append((shuffle_train_set, shuffle_test_set))
```

```python
forest = [clone(final_clf) for _ in range(n_trees)]
```

```python
accuracy_scores = []
for tree, (X_mini_train, y_mini_train) in zip(forest, subsets):
    tree.fit(X_mini_train, y_mini_train)

    y_pred = tree.predict(X_test)
    accuracy_scores.append(accuracy_score(y_test, y_pred))
```
