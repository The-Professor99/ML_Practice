#### The ff code loads the iris dataset, scales the features, then trains a linear SVM model

```python
import numpy as np
from sklearn import datasets
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
```

```python
iris = datasets.load_iris()
```

```python
X = iris["data"][:, (2, 3)] #petal length, petal width
y = (iris["target"] == 2).astype(np.float64) # Iris virginica
```

```python
svm_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("linear_svc", LinearSVC(C=1, loss="hinge"))
])
```

```python
svm_clf.fit(X, y)
```

```python
svm_clf.predict([[5.5, 1.7]])
```

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import PolynomialFeatures
```

```python
X, y = make_moons(n_samples=100, noise=0.15)
```

```python
non_polynomial_svm_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("svm_clf", LinearSVC(C=10, loss="hinge"))
])
polynomial_svm_clf = Pipeline([
    ("polynomial", PolynomialFeatures(degree=3)),
    ("scaler", StandardScaler()),
    ("svm_clf", LinearSVC(C=10, loss="hinge"))
])
```

```python
non_polynomial_svm_clf.fit(X, y)
```

```python
polynomial_svm_clf.fit(X, y)
```

```python
def plot_dataset(X, y):
    if X.shape[1] == 1:
        plt.plot(X, y, "bs")
    else:
        plt.plot(X[:, 0][y==0], X[:, 1][y==0], "bs")
        plt.plot(X[:, 0][y==1], X[:, 1][y==1], "g^")
    plt.grid(True, which='both')
    plt.xlabel(r"$x_1$", fontsize=20)
    plt.ylabel(r"$x_2$", fontsize=20, rotation=0)
```

```python
plot_dataset(X, y)
```

```python
def plot_predictions(train_X, train_y, clf, axes):
    plt.figure()
    plot_dataset(train_X, train_y)
    x0s = np.linspace(axes[0], axes[1], 100)
    x1s = np.linspace(axes[2], axes[3], 100)
    x0, x1 = np.meshgrid(x0s, x1s)
    X = np.c_[x0.ravel(), x1.ravel()]
    y_pred = clf.predict(X).reshape(x0.shape)
    y_decision = clf.decision_function(X).reshape(x0.shape)
    plt.contourf(x0, x1, y_pred, cmap=plt.cm.brg, alpha=0.2)
    plt.contourf(x0, x1, y_decision, cmap=plt.cm.brg, alpha=0.1)
```

```python
plot_predictions(X, y, non_polynomial_svm_clf, [-1.5, 2.5, -1, 1.5])
plot_predictions(X, y, polynomial_svm_clf, [-1.5, 2.5, -1, 1.5])
```

```python
from sklearn.svm import SVC
poly_kernel_svm_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("svm_clf", SVC(kernel="poly", degree=3, coef0=1, C=5))
])
```

```python
poly_kernel_svm_clf.fit(X, y)
```

```python
plot_predictions(X, y, poly_kernel_svm_clf, [-1.5, 2.5, -1, 1.5])
```

```python
rbf_svm_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf", gamma=0.1, C=7))
])
```

```python
rbf_svm_clf.fit(X, y)
```

```python
plot_predictions(X, y, rbf_svm_clf, [-1.5, 2.5, -1, 1.5])
```

#### SVMs for classification

```python
np.random.seed(42)
m = 50
X = 2 * np.random.rand(m, 1)
y = (4 + 3 * X + np.random.randn(m, 1)).ravel()
```

```python
plot_dataset(X, y)
```

```python
def plot_svm_regression2(X, y, svm_reg, axes):
    plt.figure()
    x1s = np.linspace(axes[0], axes[1], 100).reshape(100, 1)
    y_pred = svm_reg.predict(x1s)
    plt.plot(x1s, y_pred, "k-", linewidth=2, label=r"$\hat{y}$")
    plt.plot(x1s, y_pred + svm_reg.named_steps.svm_reg.epsilon, "k--")
    plt.plot(x1s, y_pred - svm_reg.named_steps.svm_reg.epsilon, "k--")
    plt.scatter(X[svm_reg.named_steps.svm_reg.support_], y[svm_reg.named_steps.svm_reg.support_], s=180, facecolors='#FFAAAA')
    plt.plot(X, y, "bo")
    plt.xlabel(r"$x_1$", fontsize=18)
    plt.legend(loc="upper left", fontsize=18)
    plt.axis(axes)

def plot_svm_regression(X, y, svm_reg, axes):
    plt.figure()
    x1s = np.linspace(axes[0], axes[1], 100).reshape(100, 1)
    y_pred = svm_reg.predict(x1s)
    plt.plot(x1s, y_pred, "k-", linewidth=2, label=r"$\hat{y}$")
    plt.plot(x1s, y_pred + svm_reg.epsilon, "k--")
    plt.plot(x1s, y_pred - svm_reg.epsilon, "k--")
    plt.scatter(X[svm_reg.support_], y[svm_reg.support_], s=180, facecolors='#FFAAAA')
    plt.plot(X, y, "bo")
    plt.xlabel(r"$x_1$", fontsize=18)
    plt.legend(loc="upper left", fontsize=18)
    plt.axis(axes)
```
