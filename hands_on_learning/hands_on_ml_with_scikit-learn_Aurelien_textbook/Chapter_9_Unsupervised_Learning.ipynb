### K-Means

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib
%matplotlib
```

```python
from sklearn.datasets import make_blobs
```

```python
blob_centers = np.array(
    [[ 0.2,  2.3],
     [-1.5 ,  2.3],
     [-2.8,  1.8],
     [-2.8,  2.8],
     [-2.8,  1.3]])
blob_std = np.array([0.4, 0.3, 0.1, 0.1, 0.1])
```

```python
X, y = make_blobs(n_samples=2000, centers=blob_centers,
                  cluster_std=blob_std, random_state=7)
```

```python
def plot_clusters(X, y=None):
    plt.scatter(X[:, 0], X[:, 1], c=y, s=1)
    plt.xlabel("$x_1$", fontsize=14)
    plt.ylabel("$x_2$", fontsize=14, rotation=0)
```

```python
plot_clusters(X)
```

```python
from sklearn.cluster import KMeans
```

```python
k = 5
kmeans = KMeans(n_clusters=k)
y_pred = kmeans.fit_predict(X)
```

```python
def plot_data(X):
    plt.plot(X[:, 0], X[:, 1], 'k.', markersize=2)

def plot_centroids(centroids, weights=None, circle_color='w', cross_color='k'):
    if weights is not None:
        centroids = centroids[weights > weights.max() / 10]
    plt.scatter(centroids[:, 0], centroids[:, 1],
                marker='o', s=30, linewidths=8,
                color=circle_color, zorder=10, alpha=0.9)
    plt.scatter(centroids[:, 0], centroids[:, 1],
                marker='x', s=50, linewidths=50,
                color=cross_color, zorder=11, alpha=1)

def plot_decision_boundaries(clusterer, X, resolution=1000, show_centroids=True,
                             show_xlabels=True, show_ylabels=True):
    mins = X.min(axis=0) - 0.1
    maxs = X.max(axis=0) + 0.1
    xx, yy = np.meshgrid(np.linspace(mins[0], maxs[0], resolution),
                         np.linspace(mins[1], maxs[1], resolution))
    Z = clusterer.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(Z, extent=(mins[0], maxs[0], mins[1], maxs[1]),
                cmap="Pastel2")
    plt.contour(Z, extent=(mins[0], maxs[0], mins[1], maxs[1]),
                linewidths=1, colors='k')
    plot_data(X)
    if show_centroids:
        plot_centroids(clusterer.cluster_centers_)

    if show_xlabels:
        plt.xlabel("$x_1$", fontsize=14)
    else:
        plt.tick_params(labelbottom=False)
    if show_ylabels:
        plt.ylabel("$x_2$", fontsize=14, rotation=0)
    else:
        plt.tick_params(labelleft=False)
```

```python
plot_decision_boundaries(kmeans, X)
```

The vast majority of the instances are clearly assigned to the appropriate cluster, but a few instances were probably misclassified. When assigning an instance to a cluster, all the K-means algorithm cares about is the distance to the centroid.

However, instead of assigning each instance to a single cluster: called hard clustering, it can be useful to give each instance a score per cluster: called soft clustering. This score can be the distance between the instance and the centroid or a similarity score(or affinity).

The transform method measures the distance from each instance to every centroid.

### Using Clustering for Preprocessing

```python
from sklearn.datasets import load_digits
```

```python
X_digits, y_digits = load_digits(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_digits, y_digits)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_reg = LogisticRegression(multi_class="ovr", solver="liblinear", random_state=42)
log_reg.fit(X_train, y_train)
```

```python
log_reg.score(X_test, y_test)
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipeline = Pipeline([
    ("kmeans", KMeans(n_clusters=50)),
    ("log_reg", LogisticRegression(multi_class="ovr", solver="liblinear", random_state=42))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
pipeline.score(X_test, y_test)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = dict(kmeans__n_clusters=range(2, 100))
```

```python
grid_clf = GridSearchCV(pipeline, param_grid, cv=3, verbose=2)
```

```python
grid_clf.fit(X_train, y_train)
```

```python
grid_clf.best_params_
```

```python
grid_clf.score(X_test, y_test)
```

```python
k = 50
kmeans = KMeans(n_clusters=k)

# fits the dataset(clusters the set into 50 clusters)
# and returns the distance from each instance to the
# centroid
X_digits_dist = kmeans.fit_transform(X_train)

# returns the index of the instance closest to the
# centroid for each cluster.
representative_digit_idx = np.argmin(X_digits_dist, axis=0)

# extracts the images of the instances gotten from above
X_representative_digits = X_train[representative_digit_idx]
```

```python
plt.figure(figsize=(8, 2))
for index, X_representative_digit in enumerate(X_representative_digits):
    plt.subplot(k // 10, 10, index + 1)
    plt.imshow(X_representative_digit.reshape(8, 8), cmap="binary", interpolation="bilinear")
    plt.axis('off')
```
