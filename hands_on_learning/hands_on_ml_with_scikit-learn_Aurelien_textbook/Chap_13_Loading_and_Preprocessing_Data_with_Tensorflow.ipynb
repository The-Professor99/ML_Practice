```python
import tensorflow as tf
```

### The Data API

```python
X = tf.range(10)
```

```python
dataset = tf.data.Dataset.from_tensor_slices(X)
```

```python
dataset
```

```python
for item in dataset:
    print(item)
```

```python
# once you have a dataset, you can apply all sorts of
# transformations to it by calling its transformation
# methods. Each method returns a new dataset, so you can
# chain transformations like this
dataset = dataset.repeat(3).batch(7)
for item in dataset:
    print(item)

# The batch method had to output a final batch of size two
# instead of seven, but you can call it with
# drop_remainder=True if you want it to drop this final
# batch so that all batches have the exact same size.
```
