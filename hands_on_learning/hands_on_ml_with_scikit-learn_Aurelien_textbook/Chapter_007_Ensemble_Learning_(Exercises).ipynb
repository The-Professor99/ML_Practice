```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
%matplotlib
```

```python
import joblib
mnist = joblib.load("datasets/mnist.pkl")
```

```python
X, y = mnist["data"], mnist["target"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train_val, X_test, y_train_val, y_test = train_test_split(X, y, test_size=10000)
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=10000)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.metrics import accuracy_score
```

```python
rfc = RandomForestClassifier()
rfc.fit(X_train, y_train)
```

```python
rfc_predictions = rfc.predict(X_val)
```

```python
accuracy_score(y_val, rfc_predictions)
```

```python
from sklearn.ensemble import ExtraTreesClassifier
```

```python
etc = ExtraTreesClassifier()
etc.fit(X_train, y_train)
```

```python
etc_predictions = etc.predict(X_val)
```

```python
accuracy_score(y_val, etc_predictions)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC()
svc.fit(X_train, y_train)
```

```python
svc_predictions = svc.predict(X_val)
```

```python
accuracy_score(y_val, svc_predictions)
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
ensemble_clf = VotingClassifier([
    ('rfclf', rfc),
    ('etclf', etc),
    ('svclf', svc)
])
ensemble_clf.fit(X_train, y_train)
```

```python
ensemble_predictions = ensemble_clf.predict(X_val)
```

```python
accuracy_score(y_val, ensemble_predictions)
```

```python
ensemble_clf.score(X_val, y_val)
```

```python
ensemble_clf.score(X_test, y_test)
```

```python
ensemble_test_predictions = ensemble_clf.predict(X_test)
```
