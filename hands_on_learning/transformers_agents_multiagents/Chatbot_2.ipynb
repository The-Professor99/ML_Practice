```python
import os
import getpass
```

```python
from langchain_cohere import ChatCohere
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults
from langgraph.checkpoint.memory import MemorySaver
from langgraph.prebuilt import create_react_agent
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage
```

```python
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_ENDPOINT"] = "https://api.smith.langchain.com"
os.environ["LANGCHAIN_PROJECT"] = "test-project"
os.environ["USER_AGENT"] = "LangChain/1.0.0"
```

```python
os.environ["LANGCHAIN_API_KEY"] = getpass.getpass()
```

```python
os.environ["COHERE_API_KEY"] = getpass.getpass()
```

```python
os.environ["TAVILY_API_KEY"] = getpass.getpass()
```

```python
model = ChatCohere(model="command-r-plus")
```

```python
# Create the agent - very simple tbvh
memory = MemorySaver()
search = TavilySearchResults(max_results=2)
tools = [search]
agent_executor = create_react_agent(model, tools, checkpointer=memory)
```

```python
username = "my_username_1"
topic = "topic_1"
```

```python
config = {
    "configurable": {
        "thread_id": f"{username}_{topic}"
    }
}
# controls what thread is run. This enables the application to support multiple
# users and multiple topics.
```

```python
def chat_with_agent(agent_executor, config):
    active = True
    while active:
        user_query = input("Message Model (Enter n to quit): ")
        if user_query.lower() == "n":
            active = False
            continue
        for event in agent_executor.stream(
            {"messages": [HumanMessage(content=user_query)]},
            stream_mode="values",
            config=config,
        ):
            message = event["messages"][-1]
            if isinstance(message, ToolMessage):
                continue
            else:
                if isinstance(message, AIMessage) and len(message.tool_calls):
                    continue
            message.pretty_print()
```

```python
chat_with_agent(agent_executor, config)
```
